# **Birth Day Problem**

**Probability that 2 person have birth day on a same date bold text**


```python
# for n =1
probability = 1.0  # person can born on any day , 365/365
print("Probability =", round(probability,2))
```

```python
n = 2

probability = 1 / 365   # for match it should match with the 1st person's birthday

print("Probability =", round(probability,2))
print("Percentage =", round(probability * 100,2), "%")
```

```python
# for diffrent number of people (Sample size )

import math

for n in [5, 23, 30, 100]:
    probability = 1 - (math.prod(365 - i for i in range(n)) / 365**n)

    print(f"n = {n}")
    print(f"Probability = {probability:.6f}")
    print(f"Percentage = {probability * 100:.2f}%")
    print()
```

# **Graph for Different Sample Size**

```python
import math
import matplotlib.pyplot as plt

n_values = [1, 2, 5, 23, 30, 50, 100, 150, 200, 250, 300, 350]

probabilities = []

for n in n_values:
    probability = 1 - (math.prod(365 - i for i in range(n)) / 365**n)
    probabilities.append(probability * 100)

# Print the results
for n, p in zip(n_values, probabilities):
    print(f"n = {n:3d}  Probability = {p:.6f}%")

# Plot
plt.figure(figsize=(10, 6))

plt.plot(
    n_values,
    probabilities,
    marker='o',
    linewidth=2,
    color='blue'
)

plt.xlabel("Number of People (n)")
plt.ylabel("Probability of Shared Birthday (%)")
plt.title("Birthday Problem: Probability of a Shared Birthday")

plt.grid(True)
plt.ylim(0, 105)

plt.show()
```
